## 계산기 1개 만들 때

```python
# calculator.py
result = 0  # 결괏값을 유지하기 위한 전역 변수

def add(num):
    global result
    result += num  # 결괏값(result)에 입력값(num) 더하기
    return result  # 결괏값 리턴

print(add(3))
print(add(4))
```

## 만약 계산기가 2개가 필요하다고..? (살면서 2개 동시에 써본 적은 없지만..ㅎ)

```python
# calculator2.py
result1 = 0
result2 = 0

def add1(num):  # 계산기1
    global result1
    result1 += num
    return result1

def add2(num):  # 계산기2
    global result2
    result2 += num
    return result2

print(add1(3))
print(add1(4))
print(add2(3))
print(add2(7))
```

## 클래스를 활용한 계산기

```python
# calculator3.py
class Calculator:
    def __init__(self):
        self.result = 0

    def add(self, num):
        self.result += num
        return self.result

cal1 = Calculator()
cal2 = Calculator()

print(cal1.add(3))
print(cal1.add(4))
print(cal2.add(3))
print(cal2.add(7))
```

## 굳이 굳이 메서드 즉, 클래스 내부에 있는 함수 첫 매개변수를 self로 지정해야 하는 이유

```python
class Person:
    name = None  # 클래스 변수 (모든 객체 공유)

    def set_name(self, name):
        Person.name = name  # 인스턴스 속성이 아니라 클래스 변수에 덮어씀

# 두 명의 객체
p1 = Person()
p2 = Person()

p1.set_name("철수")
p2.set_name("영희")

print("p1.name:", p1.name)
print("p2.name:", p2.name)
```

```python
class Person:
    name = None  # 클래스 변수 (모든 객체 공유)

    def set_name(self, name):
        self.name = name  #  인스턴스별로 저장

# 두 명의 객체
p1 = Person()
p2 = Person()

p1.set_name("철수")
p2.set_name("영희")

print("p1.name:", p1.name)  # 철수
print("p2.name:", p2.name)  # 영희
```

## 예제 2 사칙 연산 기능 계산기

```python
class FourCal:
     def setdata(self, first, second):
         self.first = first
         self.second = second
     def add(self):
         result = self.first + self.second
         return result
     def mul(self):
         result = self.first * self.second
         return result
     def sub(self):
         result = self.first - self.second
         return result
     def div(self):
         result = self.first / self.second
         return result


a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)
a.add()

# a.mul()
#
# a.sub()
#
# a.div()
#
# b.add()
#
# b.mul()
#
# b.sub()
#
# b.div()
```

## 예제 2 심화

```python
methods = ['add', 'mul', 'sub', 'div']

a = FourCal()
b = FourCal()
a.setdata(4, 2)
b.setdata(3, 8)

print("=== a 결과 ===")
for m in methods:
    print(m, ":", getattr(a, m)())  # getattr로 메서드 호출

print("=== b 결과 ===")
for m in methods:
    print(m, ":", getattr(b, m)())
```

## 예제 문제 3 강아지 멍, 고양이 야옹

```python
# 부모 클래스
class Animal:
    def __init__(self, name):
        self.name = name

    def speak(self):
        return "무슨 소리를 내야 할지 모르겠습니다."

# 자식 클래스 Dog
class Dog(Animal):
    def speak(self):  # 부모의 speak를 오버라이딩
        return "멍멍!"

# 자식 클래스 Cat
class Cat(Animal):
    def speak(self):  # 부모의 speak를 오버라이딩
        return "야옹!"

# 테스트
dog = Dog("초코")
cat = Cat("나비")

print(dog.name + "가 " + dog.speak())  # 초코가 멍멍!
print(cat.name + "가 " + cat.speak())  # 나비가 야옹!

```
